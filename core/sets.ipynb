**How Python Decides Which Item to Pop**

>Under the hood, Python sets are implemented using a hash table (a list of memory slots or "buckets"):

>Each element in the set is assigned a hash value and placed into a specific slot in memory based on that hash.

>When you call .pop(), Python simply looks for the first occupied slot in its internal hash table, removes that item, and returns it.

**Why It Can Feel Predictable (and Why You Shouldn't Rely on It)**

> You might notice that running pop() on small integers often pops the lowest number first (e.g., popping 1 from {1, 2, 3, 4}).

> This happens because for small integers, Python's internal hash function turns numbers into hash values that match themselves (e.g., hash(1) == 1).

> As a result, small integers land in the exact index slots corresponding to their value, making them appear in numerical order inside the table.

```
Caution: This is an implementation detail, not a guarantee! For strings, floats, larger numbers, or across different Python sessions (where hash randomization is enabled for security), the element popped will appear entirely random and unpredictable.
```

````python
## MAthematical Operation
set1={1,2,3,4,5,6}
set2={4,5,6,7,8,9}

### Union
union_set=set1.union(set2)
print(union_set)

## Intersection
intersection_set=set1.intersection(set2)    #creates a new set
print(intersection_set)

set1.intersection_update(set2)  #modifies the original set
print(set1)
````
