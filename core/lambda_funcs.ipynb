```python
# map is a function that applies a function to all items in a list
# map() takes two arguments - a function reference and a list

def find_cubes(a):
    return a**3

nums = [x for x in range (6)]
nums_cubes = list(map(find_cubes, nums))
print(nums_cubes)

```

```python
# passing a lambda in the map()

nums_squares = list(map(lambda x : x**2, nums))
print(nums_squares)
```

 filter() function filters out elements from an iterable to construct another iterator based on a boolean condition

 filter() accepts a bool function and an iterable to apply filter on it based on that function



```python
# given a list of employee info, find the names of all the employees whose salary is more than the avg salary in the department

emp = [
    {'name': 'Jack', 'salary': 150},
    {'name': 'Josh', 'salary': 135},
    {'name': 'Nick', 'salary': 145},
    {'name': 'Patrick', 'salary': 165},
    {'name': 'Charlie', 'salary': 200},
    {'name': 'Matilda', 'salary': 110}
]

# Approach 1 - simple list comprehension
avg = sum(e['salary'] for e in emp) / len(emp)
res = [e['name'] for e in emp if e['salary'] > avg]
print(res)

#Approach 2 - using filter and map
res_filter = list(map(lambda e: e['name'], filter(lambda e: e['salary'] > avg, emp)))
print(res_filter)
```
